## Reproduire le profilage

Depuis le dépôt `github`, utiliser Python 3.12 et le binaire DuckDB 2.0.1. Choisir un dossier temporaire disposant de plusieurs gigaoctets libres. Le script écrit les profils agrégés et efface ses partitions Parquet temporaires à la fin.

```powershell
python profiler_ticks.py --export ancien --scratch .\work_profilage --out .\rapports_profilage\ancien.json
python profiler_ticks.py --export recent --scratch .\work_profilage --out .\rapports_profilage\recent.json
```
